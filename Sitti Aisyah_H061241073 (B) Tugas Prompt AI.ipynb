## **PENDAHULUAN**  
Dalam bidang Geofisika, pemanfaatan Artificial Intelligence (AI) dapat membantu mahasiswa dalam memahami, menganalisis, dan mengolah berbagai permasalahan geosains secara lebih sistematis. Salah satu bentuk pemanfaatannya adalah melalui penyusunan prompt yang terstruktur, sehingga AI dapat digunakan sebagai alat bantu dalam proses analisis tanpa menggantikan dasar teori dan pertimbangan ilmiah.

Berikut ini disajikan beberapa contoh kasus yang mencakup empat bidang dalam Geofisika, yaitu Geofisika Padat, Sistem Informasi Geografis (SIG), Meteorologi, dan Oseanografi. Setiap kasus disusun menggunakan struktur Role, Task, Context, Input, Constraints, Output, Quality Criteria, dan Verification (R–T–C–I–K–O–Q–V). Struktur ini digunakan agar proses analisis memiliki tujuan, batasan, keluaran, dan langkah verifikasi yang jelas.


## **KASUS 1 — INTERPRETASI PENAMPANG SEISMIK 2D**

### **Role**
Kita berperan sebagai mahasiswa Geofisika yang sedang mempelajari interpretasi dasar data seismik refleksi untuk memahami kondisi dan struktur bawah permukaan.

### **Task**
Analisis penampang seismik 2D untuk mengidentifikasi reflektor utama, memperkirakan batas antar-lapisan berdasarkan pola reflektor, serta mengidentifikasi kemungkinan struktur geologi yang terlihat pada penampang.

### **Context**

Data seismik refleksi digunakan untuk memperoleh gambaran kondisi bawah permukaan berdasarkan pantulan gelombang seismik. Pantulan dapat muncul ketika gelombang melewati batas material yang memiliki perbedaan sifat fisik.

Interpretasi dilakukan dengan memperhatikan kontinuitas reflektor, bentuk, kemiringan, perubahan amplitudo, dan posisi terhadap waktu tempuh dua arah (*Two-Way Travel Time* atau TWT).

Interpretasi dalam kasus ini merupakan interpretasi awal berdasarkan karakter yang terlihat pada penampang.

### **Input**

Data yang digunakan berupa:

- Penampang seismik 2D dalam bentuk gambar;
- Sumbu horizontal berupa posisi atau jarak lintasan;
- Sumbu vertikal berupa TWT;
- Informasi lokasi atau lintasan jika tersedia.

### **Constraints**

1. Jangan menentukan jenis batuan hanya berdasarkan bentuk atau amplitudo reflektor.
2. Gunakan karakter yang benar-benar terlihat pada penampang.
3. Bedakan pengamatan langsung dengan interpretasi.
4. Jangan menyatakan struktur sebagai pasti jika bukti tidak cukup.
5. Jangan mengubah atau menghilangkan bagian data.
6. Jika reflektor tidak jelas, nyatakan sebagai ketidakpastian.
7. Jangan mengubah TWT menjadi kedalaman jika model kecepatan tidak tersedia.

### **Output**

Hasil yang direncanakan berupa:

1. Identifikasi reflektor utama;
2. Deskripsi kontinuitas dan bentuk reflektor;
3. Perkiraan batas lapisan;
4. Identifikasi kemungkinan struktur geologi;
5. Alasan yang mendukung setiap interpretasi;
6. Tabel pengamatan dan interpretasi;
7. Kesimpulan pola bawah permukaan.

### **Quality Criteria**

Hasil harus:
- Berdasarkan pola yang terlihat pada penampang;
- Mempertimbangkan kontinuitas reflektor;
- Mempertimbangkan bentuk dan kemiringan;
- Mempertimbangkan perubahan amplitudo jika terlihat;
- Mencantumkan posisi TWT yang relevan;
- Membedakan pengamatan dari interpretasi;
- Menyatakan ketidakpastian.

### **Verification**

Periksa kembali seluruh penampang setelah interpretasi. Pastikan setiap reflektor dan struktur yang disebutkan memiliki dasar visual pada data dan konsisten dengan pola reflektor di bagian penampang yang berdekatan.

### **Rencana Hasil**

Output yang diharapkan berupa deskripsi reflektor utama, kemungkinan batas lapisan, dan kemungkinan struktur bawah permukaan berdasarkan bukti yang terlihat pada penampang.

## **KASUS 2 — INTERPRETASI PENAMPANG RESISTIVITAS ERT 2D**

### **Role**

Kita berperan sebagai mahasiswa Geofisika yang sedang mempelajari interpretasi dasar data Electrical Resistivity Tomography (ERT) untuk mengetahui variasi sifat listrik bawah permukaan.

### **Task**

Analisis penampang resistivitas 2D untuk mengidentifikasi zona resistivitas tinggi dan rendah, memperkirakan batas antar-zona, serta mengidentifikasi kemungkinan anomali bawah permukaan.

### **Context**

ERT merupakan metode geofisika yang memperkirakan distribusi resistivitas bawah permukaan berdasarkan pengukuran listrik dan proses inversi. Hasilnya dapat ditampilkan sebagai penampang 2D terhadap posisi dan kedalaman.

Interpretasi ERT dipengaruhi oleh geometri pengukuran, error, regularisasi, dan model yang digunakan.

### **Input**

Data yang digunakan berupa:

- Penampang hasil inversi ERT 2D;
- Posisi sepanjang lintasan;
- Kedalaman;
- Nilai resistivitas;
- Informasi konfigurasi elektroda jika tersedia;
- Informasi lokasi survei jika tersedia.

### **Constraints**

1. Jangan menentukan jenis batuan hanya berdasarkan nilai resistivitas.
2. Gunakan pola resistivitas yang benar-benar terlihat.
3. Bedakan pengamatan dan interpretasi.
4. Jangan menyatakan anomali sebagai struktur tertentu jika bukti tidak cukup.
5. Jangan mengubah nilai resistivitas atau kedalaman.
6. Nyatakan ketidakpastian pada zona yang tidak jelas.
7. Jangan menganggap hasil inversi sebagai satu-satunya model yang mungkin.

### **Output**

Hasil yang direncanakan berupa:

1. Identifikasi zona resistivitas utama;
2. Deskripsi perubahan resistivitas;
3. Perkiraan batas antar-zona;
4. Identifikasi kemungkinan anomali;
5. Alasan setiap interpretasi;
6. Tabel pengamatan dan interpretasi;
7. Kesimpulan kondisi bawah permukaan.

### **Quality Criteria**

Hasil harus:

- Berdasarkan pola resistivitas;
- Mempertimbangkan nilai dan perubahan resistivitas;
- Mempertimbangkan posisi dan kedalaman;
- Mempertimbangkan kontinuitas zona;
- Membedakan observasi dan interpretasi;
- Menyatakan ketidakpastian.

### **Verification**

Periksa kembali seluruh penampang ERT. Pastikan setiap zona dan anomali memiliki dasar visual serta interpretasinya konsisten dengan bagian penampang di sekitarnya.

### **Rencana Hasil**

Output yang diharapkan berupa identifikasi zona resistivitas dan kemungkinan anomali bawah permukaan, tanpa menentukan litologi secara pasti.

## **KASUS 3 — PEMETAAN KERAWANAN LONGSOR MENGGUNAKAN SIG**

### **Role**

Kita berperan sebagai mahasiswa Geofisika yang sedang mempelajari pemanfaatan Sistem Informasi Geografis (SIG) untuk melakukan analisis spasial terhadap potensi kerawanan longsor.

### **Task**

Mengolah dan menggabungkan beberapa parameter spasial untuk menghasilkan peta kerawanan longsor suatu wilayah.

### **Context**

SIG dapat digunakan untuk mengelola, menganalisis, dan memvisualisasikan data yang memiliki koordinat.

Kerawanan longsor dapat dianalisis menggunakan beberapa parameter seperti elevasi, kemiringan lereng, curah hujan, penggunaan lahan, dan kondisi geologi.

Hasil analisis merupakan model spasial berdasarkan parameter yang digunakan dan bukan bukti langsung bahwa longsor pasti terjadi.

### **Input**

Data yang digunakan berupa:

- DEM;
- Data kemiringan lereng;
- Data curah hujan;
- Data penggunaan lahan;
- Data geologi;
- Batas wilayah penelitian;
- Sistem koordinat atau CRS masing-masing data.

### **Constraints**

1. Pastikan sistem koordinat data konsisten.
2. Jangan mengubah data sumber secara permanen.
3. Nyatakan parameter yang digunakan.
4. Nyatakan metode pembobotan jika menggunakan pembobotan.
5. Jangan menyatakan hasil sebagai kejadian longsor aktual.
6. Perhatikan resolusi setiap data.
7. Bedakan hasil model spasial dengan kondisi lapangan.

### **Output**

Hasil yang direncanakan berupa:

1. Peta setiap parameter;
2. Hasil pengolahan setiap parameter;
3. Peta kerawanan longsor;
4. Klasifikasi tingkat kerawanan;
5. Tabel parameter dan bobot;
6. Deskripsi distribusi kerawanan;
7. Kesimpulan hasil analisis spasial.

### **Quality Criteria**

Hasil harus:

- Menggunakan CRS yang konsisten;
- Mempertimbangkan resolusi data;
- Menjelaskan parameter yang digunakan;
- Menjelaskan metode pembobotan;
- Menunjukkan proses overlay;
- Membedakan hasil model dengan kondisi aktual.

### **Verification**

Periksa CRS, resolusi, extent, dan nilai setiap layer sebelum dilakukan overlay. Setelah peta kerawanan dibuat, periksa apakah hasil klasifikasi sesuai dengan parameter yang digunakan.

### **Rencana Hasil**

Output yang diharapkan berupa peta kerawanan longsor berbasis SIG yang menunjukkan distribusi spasial tingkat kerawanan berdasarkan parameter yang digunakan.

## **KASUS 4 — ANALISIS POLA CURAH HUJAN**


### **Role**

Kita berperan sebagai mahasiswa Geofisika yang sedang mempelajari parameter meteorologi, khususnya analisis pola curah hujan berdasarkan data pengamatan.

### **Task**

Analisis data curah hujan untuk mengetahui variasi curah hujan berdasarkan waktu, mengidentifikasi periode curah hujan tinggi dan rendah, serta melihat kemungkinan pola atau anomali pada data.

### **Context**

Curah hujan merupakan salah satu parameter meteorologi yang dapat digunakan untuk menggambarkan kondisi atmosfer dan variasi hujan suatu wilayah.

Analisis dapat dilakukan berdasarkan data harian, bulanan, atau tahunan. Interpretasi harus mempertimbangkan periode pengamatan dan kualitas data.

### **Input**

Data yang digunakan berupa:

- Data curah hujan;
- Tanggal atau waktu pengamatan;
- Satuan curah hujan;
- Lokasi stasiun;
- Informasi kelengkapan data jika tersedia.

### **Constraints**

1. Jangan mengubah data asli.
2. Jangan menganggap data kosong sebagai nilai nol.
3. Gunakan satuan yang sesuai.
4. Bedakan pengamatan dan interpretasi.
5. Jangan menentukan penyebab perubahan curah hujan hanya dari data curah hujan.
6. Tandai data yang hilang atau tidak wajar.
7. Nyatakan periode analisis.

### **Output**

Hasil yang direncanakan berupa:

1. Pola curah hujan berdasarkan waktu;
2. Periode curah hujan tinggi dan rendah;
3. Statistik dasar;
4. Grafik deret waktu;
5. Identifikasi kemungkinan anomali;
6. Tabel pengamatan dan interpretasi;
7. Kesimpulan pola curah hujan.

### **Quality Criteria**

Hasil harus:

- Berdasarkan data yang tersedia;
- Menggunakan satuan yang benar;
- Memperhatikan kelengkapan data;
- Menunjukkan periode waktu;
- Membedakan observasi dan interpretasi;
- Menyatakan keterbatasan data.

### **Verification**

Periksa kembali data sebelum dan sesudah analisis. Pastikan tidak terdapat perubahan terhadap data asli dan nilai yang dianggap anomali memang memiliki dasar dari data.

### **Rencana Hasil**

Output yang diharapkan berupa grafik dan deskripsi pola curah hujan berdasarkan waktu serta identifikasi periode tinggi, rendah, dan kemungkinan anomali.

## **KASUS 5 — ANALISIS SUHU PERMUKAAN LAUT**

### **Role**

Kita berperan sebagai mahasiswa Geofisika yang sedang mempelajari parameter oseanografi, khususnya analisis Suhu Permukaan Laut (SPL).

### **Task**

Analisis data Suhu Permukaan Laut untuk mengetahui variasi spasial dan temporal suhu serta mengidentifikasi wilayah dengan suhu relatif tinggi dan rendah.

### **Context**

Suhu Permukaan Laut merupakan salah satu parameter oseanografi yang menggambarkan kondisi termal permukaan laut.

Data SPL dapat dianalisis berdasarkan lokasi dan waktu untuk mengetahui pola distribusinya. Interpretasi perlu mempertimbangkan periode, resolusi, dan kualitas data.

### **Input**

Data yang digunakan berupa:

- Data Suhu Permukaan Laut;
- Koordinat lintang dan bujur;
- Waktu pengamatan;
- Satuan suhu;
- Informasi sumber dan resolusi data jika tersedia.

### **Constraints**

1. Jangan mengubah nilai data asli.
2. Gunakan satuan suhu yang sesuai.
3. Bedakan variasi spasial dan temporal.
4. Jangan menentukan penyebab perubahan suhu tanpa data pendukung.
5. Jangan mengisi data kosong tanpa metode yang jelas.
6. Pertahankan informasi koordinat.
7. Nyatakan periode dan resolusi data.

### **Output**

Hasil yang direncanakan berupa:

1. Distribusi SPL secara spasial;
2. Perubahan SPL berdasarkan waktu;
3. Identifikasi wilayah suhu relatif tinggi dan rendah;
4. Peta distribusi SPL;
5. Grafik perubahan SPL;
6. Tabel pengamatan dan interpretasi;
7. Kesimpulan mengenai pola SPL.

### **Quality Criteria**

Hasil harus:

- Berdasarkan nilai SPL;
- Mempertimbangkan posisi geografis;
- Mempertimbangkan waktu pengamatan;
- Menggunakan skala suhu yang sesuai;
- Membedakan observasi dan interpretasi;
- Menyatakan keterbatasan resolusi atau data.

### **Verification**

Periksa kembali peta dan data SPL. Pastikan pola suhu yang diidentifikasi sesuai dengan nilai data dan posisi geografisnya. Periksa pula adanya data kosong atau nilai yang tidak wajar.

### **Rencana Hasil**

Output yang diharapkan berupa peta dan grafik distribusi Suhu Permukaan Laut serta deskripsi wilayah dengan suhu relatif tinggi dan rendah.
